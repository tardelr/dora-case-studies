```python
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer
from peft import PeftModel
import lm_eval
from lm_eval.models.huggingface import HFLM
```

```python
BASE_MODEL_ID = "Qwen/Qwen2.5-3B"
OUTPUT_DIR    = "./qwen-3b-dora"
ADAPTER_PATH  = f"{OUTPUT_DIR}/final_adapter"
TASKS         = ["hellaswag", "arc_easy"]
NUM_FEWSHOT   = 0
LIMIT         = None          # set to e.g. 50 for a quick sanity check
```

```python
tokenizer = AutoTokenizer.from_pretrained(BASE_MODEL_ID)
base_model = AutoModelForCausalLM.from_pretrained(
    BASE_MODEL_ID,
    device_map="auto",
    torch_dtype=torch.bfloat16,
)
```

```python
# ── Load & merge DoRA adapter ────────────────────────────────
model = PeftModel.from_pretrained(base_model, ADAPTER_PATH)
merged_model = model.merge_and_unload()
print("DoRA adapter merged successfully.")
```

```python
eval_model = HFLM(
    pretrained=merged_model,
    tokenizer=tokenizer,
    batch_size="auto",
)
```

```python
results = lm_eval.simple_evaluate(
    model=eval_model,
    tasks=TASKS,
    num_fewshot=NUM_FEWSHOT,
    limit=LIMIT,
)
```

```python
for task, metrics in results["results"].items():
    acc      = metrics.get("acc,none",      metrics.get("acc", "N/A"))
    acc_norm = metrics.get("acc_norm,none", metrics.get("acc_norm", "N/A"))
    print(f"{task:>12s}  acc={acc}  acc_norm={acc_norm}")
```

```python
import json
with open('results.json', 'w') as f: 
    json.dump(results, f, default=str)
```
